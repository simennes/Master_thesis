```python
# sample 500 times from normal distribution with mean 0 and standard deviation 1, and plot the histogram of the samples
import numpy as np
import matplotlib.pyplot as plt
samples = np.random.normal(20, 3, 1000)
plt.hist(samples, bins=30, density=True)
plt.title('Histogram of 1000 samples from N(20,3)')
plt.xlabel('Phenotype value')
plt.ylabel('Density')
#remove box around the plot
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
